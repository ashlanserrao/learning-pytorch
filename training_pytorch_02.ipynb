```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
```

```python
df.shape
```

```python
df.drop(columns=['id', 'Unnamed: 32'], inplace=True)
```

```python
df.head()
```

#### Train Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 1:], df.iloc[:, 0], test_size=0.2)
```

### scaling

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train
```

```python
y_train
```

### Label Encoding

```python
# Retrieve fresh string splits using y_test's index to restore original labels
y_train_series = df.loc[~df.index.isin(y_test.index), 'diagnosis']
y_test_series = df.loc[y_test.index, 'diagnosis']

encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train_series)
y_test = encoder.transform(y_test_series)
```

```python
y_train
```

### Numpy to pytorch tensors

```python
X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)
```

```python
X_train_tensor.shape
```

```python
from torch.utils.data import Dataset, DataLoader
class CustomDataset(Dataset):

  def __init__(self, features, labels):
    self.features = features
    self.labels = labels

  def __len__(self):
    return self.features.shape[0]

  def __getitem__(self, index):
    return self.features[index], self.labels[index]

```

```python
train_dataset = CustomDataset(X_train_tensor, y_train_tensor)
test_dataset = CustomDataset(X_test_tensor, y_test_tensor)
```

```python
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=True)
```

### Important parameters


```python
learning_rate = 0.1
epochs = 25
```

```python
# define loss function
loss_function = nn.BCELoss()
```

### Training Pipeline

```python
import torch
import torch.nn as nn

# Redefine the corrected MySimpleNN class
class MySimpleNN(nn.Module):
  def __init__(self, num_features):
    super().__init__()
    self.linear = nn.Linear(num_features, 1, dtype=torch.float64)
    self.sigmoid = nn.Sigmoid()

  def forward(self, X):
    out = self.linear(X)
    out = self.sigmoid(out)
    return out


# Create model
model = MySimpleNN(X_train_tensor.shape[1])

# define optimizer
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)

# Ensure data is in float64 to match model weights
X_train_tensor = X_train_tensor.to(torch.float64)
y_train_tensor = y_train_tensor.to(torch.float64)

```

```python
model.linear.bias
```

```python
# define a loop
for epoch in range(epochs):
  for batch_features, batch_labels in train_loader:

    # forward pass
    y_pred = model(batch_features)

    # loss calculation - cast batch_labels to double (float64) to match y_pred
    loss = loss_function(y_pred, batch_labels.view(-1, 1).to(torch.float64))

    # clear gradients
    optimizer.zero_grad()

    # backward pass
    loss.backward()

    # parameter update
    optimizer.step()

    # printing loss in every epoch
    print(f"Epoch: {epoch+1}, loss: {loss.item()}")
```

## Evaluation

```python
# model evaluation using test_loader
model.eval()
accuracy_list = []

with torch.no_grad():
  for batch_features, batch_labels in test_loader:
    y_pred = model.forward(batch_features)
    y_pred = (y_pred > 0.8).float()

    batch_accuracy = (y_pred.view(-1) == batch_labels).float().mean().item()
    accuracy_list.append(batch_accuracy)

overall_accuracy = sum(accuracy_list)/len(accuracy_list)
print(f"Accuracy: {overall_accuracy:.4f}")
```
